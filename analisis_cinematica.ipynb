#Definiciοn de matrices

```python
import sympy as sp

# 1. Definiciones símbolicas y fijas
# Variables activas
q1, q2, q3, q4, q5, q6 = sp.symbols('q1 q2 q3 q4 q5 q6')

# Distancias en cm
L1 = 2.2
L2 = 29.6
L3 = 3
L4 = 30
L5 = 2
L6 = 2

# -----------------------------------------------------------

# 2. PANEL DE EQUIVALENCIAS

# Para cada eslabón, definimos sus parámetros DH (theta, d, a, alpha).
# 90 grados =  pi/2 radianes = sp.pi/2
# -90 grados = -pi/2 radianes = -sp.pi/2

# Eslabón 1
theta1 = q1
d1 = L1
a1 = 0
alpha1 = sp.pi/2  # Usamos sp.pi para que SymPy sepa que es el número Pi

# Eslabón 2
theta2 = q2
d2 = 0
a2 = L2
alpha2 = sp.pi/2

# Eslabón 3
theta3 = q3
d3 = L3
a3 = 0
alpha3 = -sp.pi/2

# Eslabón 4
theta4 = q4
d4 = 0
a4 = L4
alpha4 = -sp.pi/2

# Eslabón 5
theta5 = q5
d5 = L5
a5 = 0
alpha5 = sp.pi/2

# Eslabón 6
theta6 = q6
d6 = 0
a6 = 0
alpha6 = -sp.pi/2

# -----------------------------------------------------------

# 3. CREACIÓN DE MATRICES (Plantillas que absorben las equivalencias)

T1 = sp.Matrix([
    [sp.cos(theta1), -sp.sin(theta1)*sp.cos(alpha1),  sp.sin(theta1)*sp.sin(alpha1), a1*sp.cos(theta1)],
    [sp.sin(theta1),  sp.cos(theta1)*sp.cos(alpha1), -sp.cos(theta1)*sp.sin(alpha1), a1*sp.sin(theta1)],
    [0,               sp.sin(alpha1),                sp.cos(alpha1),                d1],
    [0,               0,                             0,                             1]
])

T2 = sp.Matrix([
    [sp.cos(theta2), -sp.sin(theta2)*sp.cos(alpha2),  sp.sin(theta2)*sp.sin(alpha2), a2*sp.cos(theta2)],
    [sp.sin(theta2),  sp.cos(theta2)*sp.cos(alpha2), -sp.cos(theta2)*sp.sin(alpha2), a2*sp.sin(theta2)],
    [0,               sp.sin(alpha2),                sp.cos(alpha2),                d2],
    [0,               0,                             0,                             1]
])

T3 = sp.Matrix([
    [sp.cos(theta3), -sp.sin(theta3)*sp.cos(alpha3),  sp.sin(theta3)*sp.sin(alpha3), a3*sp.cos(theta3)],
    [sp.sin(theta3),  sp.cos(theta3)*sp.cos(alpha3), -sp.cos(theta3)*sp.sin(alpha3), a3*sp.sin(theta3)],
    [0,               sp.sin(alpha3),                sp.cos(alpha3),                d3],
    [0,               0,                             0,                             1]
])

T4 = sp.Matrix([
    [sp.cos(theta4), -sp.sin(theta4)*sp.cos(alpha4),  sp.sin(theta4)*sp.sin(alpha4), a4*sp.cos(theta4)],
    [sp.sin(theta4),  sp.cos(theta4)*sp.cos(alpha4), -sp.cos(theta4)*sp.sin(alpha4), a4*sp.sin(theta4)],
    [0,               sp.sin(alpha4),                sp.cos(alpha4),                d4],
    [0,               0,                             0,                             1]
])

T5 = sp.Matrix([
    [sp.cos(theta5), -sp.sin(theta5)*sp.cos(alpha5),  sp.sin(theta5)*sp.sin(alpha5), a5*sp.cos(theta5)],
    [sp.sin(theta5),  sp.cos(theta5)*sp.cos(alpha5), -sp.cos(theta5)*sp.sin(alpha5), a5*sp.sin(theta5)],
    [0,               sp.sin(alpha5),                sp.cos(alpha5),                d5],
    [0,               0,                             0,                             1]
])

T6 = sp.Matrix([
    [sp.cos(theta6), -sp.sin(theta6)*sp.cos(alpha6),  sp.sin(theta6)*sp.sin(alpha6), a6*sp.cos(theta6)],
    [sp.sin(theta6),  sp.cos(theta6)*sp.cos(alpha6), -sp.cos(theta6)*sp.sin(alpha6), a6*sp.sin(theta6)],
    [0,               sp.sin(alpha6),                sp.cos(alpha6),                d6],
    [0,               0,                             0,                             1]
])

# 4. Visualización
print("--- MATRIZ 1 ---")
sp.pprint(T1, use_unicode=True)
print("\n--- MATRIZ 2 ---")
sp.pprint(T2, use_unicode=True)
print("\n--- MATRIZ 3 ---")
sp.pprint(T3, use_unicode=True)
print("\n--- MATRIZ 4 ---")
sp.pprint(T4, use_unicode=True)
print("\n--- MATRIZ 5 ---")
sp.pprint(T5, use_unicode=True)
print("\n--- MATRIZ 6 ---")
sp.pprint(T6, use_unicode=True)
```

```python
TH = T1 @ T2
TH = sp.simplify(TH)
print(sp.sstr(TH))
#--------------------
TH = TH @ T3
TH = sp.simplify(TH)
print(sp.sstr(TH))
#--------------------
TH = TH @ T4
TH = sp.simplify(TH)
print(sp.sstr(TH))
#--------------------
TH = TH @ T5
TH = sp.simplify(TH)
print(sp.sstr(TH))
#--------------------
TH = TH @ T6
TH = sp.simplify(TH)
print(sp.sstr(TH))
```

```python
TH_final = sp.simplify(TH)
print(" Matriz Final - Transformación homogénea")
print(sp.sstr(TH_final))
```

#Puntos

```python
import math

x = TH_final[0,3]
y = TH_final[1,3]
z = TH_final[2,3]

print(f"x = {x}")
print(f"y = {y}")
print(f"z = {z}")

# Usando la función especializada
rad1 = math.radians(30)
rad2 = math.radians(70)

# Sustitución de valores numéricos
# (SymPy espera radianes o sp.pi)
P_x = x.subs({q1: rad1, q2: rad2}).evalf()
P_y = y.subs({q1: rad1, q2: rad2}).evalf()

print(f"X: {P_x}, Y: {P_y}")
```

#Matriz Jacobiana

####Directa

```python
x = TH_final[0,3]
y = TH_final[1,3]
z = TH_final[2,3]

# Vector posición
pos = sp.Matrix([x, y, z])

Jv = pos.jacobian([q1, q2, q3, q4, q5, q6])

Jv_s = sp.simplify(Jv)

print(sp.sstr(Jv_s))
```

####Paso a paso

```python
# 1. Extracción de componentes de posición (Corregido: columna 3 es la cuarta)
x = TH_final[0,3]
y = TH_final[1,3]
z = TH_final[2,3]

# 2. Cálculo de derivadas parciales una por una
# Fila 1: Derivadas de X respecto a cada articulación
j11 = sp.diff(x, q1)
j12 = sp.diff(x, q2)
j13 = sp.diff(x, q3)
j14 = sp.diff(x, q4)
j15 = sp.diff(x, q5)
j16 = sp.diff(x, q6)

# Fila 2: Derivadas de Y respecto a cada articulación
j21 = sp.diff(y, q1)
j22 = sp.diff(y, q2)
j23 = sp.diff(y, q3)
j24 = sp.diff(y, q4)
j25 = sp.diff(y, q5)
j26 = sp.diff(y, q6)

# Fila 3: Derivadas de Z respecto a cada articulación
j31 = sp.diff(z, q1)
j32 = sp.diff(z, q2)
j33 = sp.diff(z, q3)
j34 = sp.diff(z, q4)
j35 = sp.diff(z, q5)
j36 = sp.diff(z, q6)

derivadas = [
    [j11, j12, j13, j14, j15, j16], # Fila de X
    [j21, j22, j23, j24, j25, j26], # Fila de Y
    [j31, j32, j33, j34, j35, j36]  # Fila de Z
]

nombres_ejes = ["X", "Y", "Z"]

print("--- Listado de Derivadas Parciales ---")
for i, fila in enumerate(derivadas):
    eje = nombres_ejes[i]
    for j, valor in enumerate(fila):
        # Imprime el nombre de la derivada (ej: dX/dq1) y su valor simplificado
        print(f"d{eje} / dq{j+1}:")
        sp.pprint(sp.simplify(valor))
        print("-" * 20)

# 3. Ensamblaje manual de la Matriz Jacobiana
Jv_m = sp.Matrix([
    [j11, j12, j13, j14, j15, j16],
    [j21, j22, j23, j24, j25, j26],
    [j31, j32, j33, j34, j35, j36]
])

# 4. Simplificación y visualización
Jv_ms = sp.simplify(Jv_m)

print("--- Jacobiano calculado elemento por elemento ---")
sp.pprint(Jv_ms)
```

#Cálculo de singularidad

```python
import itertools

# Aseguramos que Jv_s sea la matriz de tu robot antes de esto
m, n = Jv_s.shape

if m == n:
    print("--- Robot Determinado (Matriz Cuadrada) ---")
    detJ = sp.factor(sp.trigsimp(Jv_s.det()))
    print("det(J) simbólico =")
    print(detJ)

    print("-" * 50)
    print("Resolviendo la ecuación de singularidad (det(J) = 0)...")

    variables_cuadrado = [q1, q2, q3, q4, q5, q6]  

    try:
        # Igualando a cero (sp.solve asume = 0 por defecto)
        soluciones_cuadrado = sp.solve(detJ, variables_cuadrado)

        print("\n=== POSTURAS DE SINGULARIDAD ENCONTRADAS ===")
        if not soluciones_cuadrado:
            print("El solucionador terminó, pero no encontró singularidades.")
        else:
            sp.pprint(soluciones_cuadrado, use_unicode=True)

    except Exception as e:
        print(f"\nEl cálculo simbólico falló o excedió los límites: {e}")

elif m < n:
    print("--- Robot Redundante (Análisis de Menores) ---")

    # 1. Creamos una lista para almacenar nuestras ecuaciones
    ecuaciones_singularidad = []

    # 1. Obtenemos todas las combinaciones posibles de columnas para hacer matrices m x m
    # Si es 3x4, buscará todas las formas de agrupar 3 columnas.
    indices_columnas = list(range(n))
    combinaciones = list(itertools.combinations(indices_columnas, m))

    print(f"Se encontraron {len(combinaciones)} submatrices cuadradas ({m}x{m}) posibles.\n")

    # 2. Iteramos sobre cada combinación, extraemos la submatriz y sacamos su determinante
    for i, cols in enumerate(combinaciones):

        # Extraemos solo las columnas de esta combinación (¡sin modificar la original!)
        J_menor = Jv_s[:, cols]

        # Calculamos y simplificamos el determinante
        det_sub = sp.simplify(J_menor.det())

        # Guardamos la ecuación en la lista
        ecuaciones_singularidad.append(det_sub)

        print(f"Ecuación {i+1} (Submatriz con columnas {cols}):")
        print(f"det_{i+1} = {det_sub} = 0\n")

    print("-" * 50)
    print("La singularidad ocurre cuando TODOS los determinantes anteriores son simultáneamente CERO.")

else:
    print("--- Robot Sobre-restringido (m > n) ---")
    print("El robot tiene menos grados de libertad que las dimensiones espaciales que intentas controlar.")
```

```python
## Calcular si. -> No es una matriz cuadrada             ##
#               -> No son determinantes muy complejas

print("-" * 50)
print("Resolviendo el sistema. Esto puede tardar dependiendo de la complejidad...")

# 2. Definimos qué variables queremos despejar (las de tu robot)
variables = [q1, q2, q3, q4, q5, q6]

# 3. sp.solve toma una lista de expresiones (asume que están igualadas a 0)
# y busca los valores de 'variables' que cumplen todas a la vez.
try:
    soluciones = sp.solve(ecuaciones_singularidad, variables)

    print("\n=== POSTURAS DE SINGULARIDAD ENCONTRADAS ===")
    if not soluciones:
        print("El solucionador terminó, pero no encontró posturas donde todos los determinantes sean cero a la vez (raro, pero posible en ciertos diseños).")
    else:
        sp.pprint(soluciones, use_unicode=True)

except Exception as e:
    print(f"\nEl cálculo simbólico falló o excedió los límites de SymPy: {e}")
```

```python
import numpy as np

print("--- Inicializando el Controlador Numérico ---")

# 1. Empaquetar variables (Asegúrate de poner TODAS las variables de tu Jacobiano)
variables_simbolicas = (q1, q2, q3, q4, q5, q6)

# 2. LAMBDIFY: El puente entre SymPy y NumPy.
# Convierte tu matriz simbólica Jv_s en una función numérica ultrarrápida.
Jacobiano_Numerico = sp.lambdify(variables_simbolicas, Jv_s, "numpy")

print("Función Jacobiana numérica generada con éxito.")
print("-" * 50)

# ==========================================
# BUCLE SIMULADO DE CONTROL EN TIEMPO REAL
# ==========================================

# 3. Imagina que los sensores del robot leen estas posiciones actuales:
# q1 = 0 rad, q2 = 45° (0.785 rad), q3 = 0 rad, q4 = 0.5 metros, q5 = 0 rad, q6 = 0 rad
valores_q_actuales = [0.0, 0.785, 0.0, 0.5, 0.0, 0.0]

print(f"Valores actuales de las articulaciones: {valores_q_actuales}")

# 4. Evaluamos el Jacobiano con los números reales
# El asterisco (*) desempaqueta la lista para pasar cada valor a la función
J_evaluada = Jacobiano_Numerico(*valores_q_actuales)

print("=== CÁLCULO PASO A PASO DE LA PSEUDOINVERSA ===\n")

print("--- 0. Matriz Jacobiana Evaluada (J) ---")
print(np.round(J_evaluada, 4))

# Paso 1: Transpuesta del Jacobiano (J^T)
J_T = np.transpose(J_evaluada)
print("\n--- 1. Matriz Transpuesta (J^T) ---")
print(np.round(J_T, 4))

# Paso 2: Multiplicación J * J^T (Matriz Gramiana)
# En NumPy, el operador '@' se usa para la multiplicación de matrices pura
J_JT = J_evaluada @ J_T
print("\n--- 2. Gramiana (J * J^T) ---")
print(np.round(J_JT, 4))

# Paso 3: Inversa de la Gramiana (J * J^T)^-1
# Como J_JT es cuadrada (3x3), podemos usar la inversa clásica linalg.inv
try:
    inv_J_JT = np.linalg.inv(J_JT)
    print("\n--- 3. Inversa de la Gramiana (J * J^T)^-1 ---")
    print(np.round(inv_J_JT, 4))

    # Paso 4: Producto final para obtener la Pseudoinversa
    J_pseudoinversa_manual = J_T @ inv_J_JT
    print("\n--- 4. Pseudoinversa Final J^T * (J * J^T)^-1 ---")
    print(np.round(J_pseudoinversa_manual, 4))

    # Extra: Verificación rápida con la función interna de NumPy
    print("\n--- EXTRA: Verificación con np.linalg.pinv(J) ---")
    print(np.round(np.linalg.pinv(J_evaluada), 4))
    print("(Si ambos resultados son iguales, ¡la matemática es perfecta!)")

except np.linalg.LinAlgError:
    # Este bloque te salva si (J * J^T) no tiene inversa (¡Estás en una singularidad!)
    print("\n[ERROR MATEMÁTICO]")
    print("La matriz (J * J^T) es singular y no se puede invertir.")
    print("El robot está en una postura crítica. Aquí es donde la división por cero destruiría el programa.")
```
